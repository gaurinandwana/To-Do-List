```python
import pandas as pd
import os

FILE_NAME = "todo_list.csv"

# Create the file if it doesn't exist
if not os.path.exists(FILE_NAME):
    df = pd.DataFrame(columns=["task", "status"])
    df.to_csv(FILE_NAME, index=False)

print("✅ To-Do List is ready!")
```

```python
def add_task(task):
    df = pd.read_csv(FILE_NAME)

    new_task = pd.DataFrame({
        "task": [task],
        "status": ["Pending"]
    })

    df = pd.concat([df, new_task], ignore_index=True)
    df.to_csv(FILE_NAME, index=False)

    print(f"✅ Task added: {task}")
```

```python
def view_tasks():
    df = pd.read_csv(FILE_NAME)

    if df.empty:
        print("\n📋 No tasks found!")
        return

    print("\n📋 YOUR TO-DO LIST")
    print("-" * 40)

    for index, row in df.iterrows():
        if row["status"] == "Completed":
            symbol = "✅"
        else:
            symbol = "⬜"

        print(f"{index + 1}. {symbol} {row['task']} - {row['status']}")
```

```python
def complete_task(task_number):
    df = pd.read_csv(FILE_NAME)

    if task_number < 1 or task_number > len(df):
        print("❌ Invalid task number!")
        return

    df.loc[task_number - 1, "status"] = "Completed"
    df.to_csv(FILE_NAME, index=False)

    print("🎉 Task marked as completed!")
```

```python
def delete_task(task_number):
    df = pd.read_csv(FILE_NAME)

    if task_number < 1 or task_number > len(df):
        print("❌ Invalid task number!")
        return

    deleted_task = df.loc[task_number - 1, "task"]

    df = df.drop(task_number - 1).reset_index(drop=True)
    df.to_csv(FILE_NAME, index=False)

    print(f"🗑️ Deleted: {deleted_task}")
```

```python
def show_statistics():
    df = pd.read_csv(FILE_NAME)

    total = len(df)
    completed = len(df[df["status"] == "Completed"])
    pending = total - completed

    print("\n📊 TASK STATISTICS")
    print("-" * 30)
    print(f"Total tasks     : {total}")
    print(f"Completed tasks : {completed}")
    print(f"Pending tasks   : {pending}")

    if total > 0:
        percentage = (completed / total) * 100
        print(f"Completion rate : {percentage:.1f}%")
```

```python
def todo_list():
    while True:
        print("\n" + "=" * 40)
        print("📝 TO-DO LIST")
        print("=" * 40)

        print("1. Add Task")
        print("2. View Tasks")
        print("3. Complete Task")
        print("4. Delete Task")
        print("5. View Statistics")
        print("6. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            task = input("Enter task: ")
            add_task(task)

        elif choice == "2":
            view_tasks()

        elif choice == "3":
            view_tasks()
            try:
                number = int(input("\nEnter task number to complete: "))
                complete_task(number)
            except ValueError:
                print("❌ Please enter a valid number.")

        elif choice == "4":
            view_tasks()
            try:
                number = int(input("\nEnter task number to delete: "))
                delete_task(number)
            except ValueError:
                print("❌ Please enter a valid number.")

        elif choice == "5":
            show_statistics()

        elif choice == "6":
            print("\n👋 Goodbye! Stay productive!")
            break

        else:
            print("❌ Invalid choice. Please try again.")


todo_list()
```
